# 00 — Configuração do ambiente
Roda **igual no VS Code e no Colab**. Em ambos, o código compartilhado fica em `src/` e os notebooks só o chamam.

- **VS Code**: crie o venv, `pip install -r requirements.txt`, escolha o kernel e rode.
- **Colab**: ajuste `REPO_URL` abaixo e rode a célula de bootstrap (clona o repo, instala dependências e entra na pasta).
- Para o Git não carregar saídas de notebook, cada pessoa roda **uma vez**: `nbstripout --install`.

```python
# --- BOOTSTRAP DO COLAB (no VS Code esta célula não faz nada) ---
import os, subprocess, sys
from pathlib import Path

IN_COLAB = "google.colab" in sys.modules
REPO_URL = "https://github.com/<ORG_OU_USUARIO>/Projeto_Churn_Telecom.git"   # TODO: trocar pela URL do repositório do grupo
REPO_NAME = "Projeto_Churn_Telecom"

if IN_COLAB:
    dest = Path("/content") / REPO_NAME
    if not dest.exists():
        subprocess.run(["git", "clone", REPO_URL, str(dest)], check=True)
    else:
        subprocess.run(["git", "-C", str(dest), "pull"], check=True)   # pega o que o grupo atualizou
    os.chdir(dest)
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-r", "requirements.txt"], check=True)
print("Colab:", IN_COLAB, "| cwd:", Path.cwd())
```

```python
# Localiza a raiz do repositório (funciona no VS Code e no Colab) e a coloca no sys.path
import sys
from pathlib import Path
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "src" / "config.py").exists())
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
print("Raiz do projeto:", ROOT)
```
